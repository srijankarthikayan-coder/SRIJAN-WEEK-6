```python
df["Gender"] = df["Gender"].str.strip() #remove space from begining to end
df["Gender"] = df["Gender"].str.title() #capitalization
```

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"],
                                         errors="coerce",
                                         )
```

```python
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"],
                                         errors="coerce",
                                         )
```

```python
#Admission analysis & Length of stay

df["Admission Type"].value_counts()
```

```python
#calculate percentages

admission_percentage = (
    df["Admission Type"].value_counts(normalize=True)*100
)
admission_percentage
```

```python
df["Length of Stay"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days
```

```python
#Stay Categories

bins = [-1, 3, 7, float("inf")]

labels = [
    "Short Stay",
    "Medium Stay",
    "Long Stay"
]

df["Stay Category"] = pd.cut(
    df["Length of Stay"],
    bins=bins,
    labels=labels
)
df["Stay Category"]
```

```python
df["Admission Year"] = df["Date of Admission"].dt.year
```

```python
df["Admission Month"] = df["Date of Admission"].dt.month
```

```python
df[
    [
        "Date of Admission",
        "Admission Year",
        "Admission Month"
    ]
].head()
```

```python
#Count Monthly Admissions

monthly_admissions = (
    df.groupby("Admission Month")
      .size()
      .reset_index(name="Admissions")
)
```

```python
#Admission Trend

sns.lineplot(
    data=monthly_admissions,
    x="Admission Month",
    y="Admissions",
    marker="o"
)

plt.title("Monthly Admissions Trend")
plt.xlabel("Month")
plt.ylabel("No of Admissions")

plt.xticks(rotation=45)

plt.show()
```

```python
#compare admission types over time

monthly_admission_type = (
    df.groupby(
        ["Admission Month", "Admission Type"]
    )
    .size()
    .reset_index(name="Admissions")
    )
```

```python
monthly_admission_type.head()
```

```python
sns.lineplot(
    data=monthly_admission_type,
    x="Admission Month",
    y="Admissions",
    hue="Admission Type",
    marker="o"
)

plt.title("Monthly Admission Type Trend")
plt.xlabel("Month")
plt.ylabel("No of Admissions")

plt.xticks(rotation=45)

plt.show()
```

```python
numeric_clos=[
    "Age",
    "Billing Amount",
    "Length of Stay"
]

corr=df[numeric_clos].corr()
corr
```

```python
plt.figure(figsize=(7,5))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Healthcare variables correlation matrix")
plt.show()
```
